```python
def missing_report(df: pd.DataFrame) -> pd.DataFrame:
    rep = pd.isna(df).mean().to_frame(name="missing_rate")
    rep["dtype"] = df.dtypes.astype(str)
    return rep.sort_values("missing_rate", ascending=False)
```

```python
app_train_missing = missing_report(application_train)
app_train_missing = app_train_missing.reset_index().rename(columns={"index": "column_name"})
app_train_missing
```
